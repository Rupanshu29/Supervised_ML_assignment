```python
#Task 1: Accuracy, precision, recall and F1 from the spam filter confusion matrix
tp = 50
tn = 120
fp = 10
fn = 20

total = tp + tn + fp + fn
acc = (tp + tn) / total
prec = tp / (tp + fp)
rec = tp / (tp + fn)
f1 = 2 * prec * rec / (prec + rec)

print("Accuracy = (TP + TN) / total =", tp + tn, "/", total, "=", round(acc, 3))
print("Precision = TP / (TP + FP) =", tp, "/", tp + fp, "=", round(prec, 3))
print("Recall = TP / (TP + FN) =", tp, "/", tp + fn, "=", round(rec, 3))
print("F1 = 2 * P * R / (P + R) =", round(f1, 3))
```

```python
#Task 2: Function to print a confusion matrix
def print_confusion_matrix(tp, tn, fp, fn):
    print("                    Predicted Positive    Predicted Negative")
    print("Actual Positive     " + ("TP = " + str(tp)).ljust(22) + "FN = " + str(fn))
    print("Actual Negative     " + ("FP = " + str(fp)).ljust(22) + "TN = " + str(tn))

print_confusion_matrix(50, 120, 10, 20)
print()
print_confusion_matrix(40, 150, 20, 10)
```

```python
#Task 3: Precision and recall for the fake news model
tp = 60
fn = 30
tn = 120
fp = 10

prec = tp / (tp + fp)
rec = tp / (tp + fn)

print_confusion_matrix(tp, tn, fp, fn)
print("\nPrecision:", round(prec, 3))
print("Recall:", round(rec, 3))

# Precision 0.857 means about 86 of every 100 posts flagged as fake were really fake.
# Recall 0.667 means the model found only two thirds of all the fake posts.
# Note: the task says 80 posts were flagged, but TP + FP = 60 + 10 = 70. I used TP, FN, TN and FP as given.
# If 80 posts were flagged, FP would be 20 and the precision would be 0.75.
```

```python
#Task 4: F1 score for the food delivery fraud detection system
tp = 40
tn = 150
fp = 20
fn = 10

prec = tp / (tp + fp)
rec = tp / (tp + fn)
f1 = 2 * (prec * rec) / (prec + rec)

print("Precision:", round(prec, 3))
print("Recall:", round(rec, 3))
print("F1 score:", round(f1, 3))

# The F1 score (0.727) balances precision and recall: the model catches most frauds (recall 0.8), but one in three of its fraud alerts is wrong (precision 0.667).
```
