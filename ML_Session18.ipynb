```python
#Task 1: 5-fold cross-validation with KNN on Iris
import numpy as np
from sklearn.datasets import load_iris, load_wine, load_breast_cancer
from sklearn.model_selection import cross_val_score, StratifiedKFold, KFold
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

iris = load_iris()
scores = cross_val_score(KNeighborsClassifier(), iris.data, iris.target, cv=5)

print("Scores of each fold:", scores.round(3))
print("Average accuracy:", round(scores.mean(), 3))
```

```python
#Task 2: 10-fold cross-validation with a decision tree on the breast cancer dataset
bc = load_breast_cancer()
sc10 = cross_val_score(DecisionTreeClassifier(random_state=1), bc.data, bc.target, cv=10)

for i in range(len(sc10)):
    print("Fold", i + 1, "accuracy:", round(sc10[i], 3))
print("Mean accuracy:", round(sc10.mean(), 3))
```

```python
#Task 3: StratifiedKFold on Iris and class counts in each fold
print("Whole dataset class counts:", np.bincount(iris.target))

print("\nStratifiedKFold (class counts in each test fold):")
skf = StratifiedKFold(n_splits=5)
for i, (tr, te) in enumerate(skf.split(iris.data, iris.target)):
    print("Fold", i + 1, "->", np.bincount(iris.target[te]))

print("\nNormal KFold without stratification, for comparison:")
kf = KFold(n_splits=5)
for i, (tr, te) in enumerate(kf.split(iris.data, iris.target)):
    print("Fold", i + 1, "->", np.bincount(iris.target[te], minlength=3))

# StratifiedKFold keeps the same class ratio in every fold (10, 10, 10 in each one).
# The normal KFold takes the rows in order, and Iris is sorted by class, so some folds have only one class. That would give a bad and unfair test.
```

```python
#Task 4: Compare cv = 3, 5 and 10 for RandomForest on the Wine dataset
wine = load_wine()
rf = RandomForestClassifier(n_estimators=100, random_state=1)

for k in [3, 5, 10]:
    s = cross_val_score(rf, wine.data, wine.target, cv=k)
    print("cv =", k, "-> mean:", round(s.mean(), 3), "std:", round(s.std(), 4))

# The standard deviation is almost the same for all three (about 0.027). By the numbers cv = 3 is the lowest, but the difference is too small to matter.
# cv = 5 and cv = 10 give a higher mean accuracy (0.978) than cv = 3 (0.95), because each fold trains on more data.
# So cv = 5 is a good choice: stable scores, high accuracy and less running time than cv = 10.
```

```python
#Task 5: Function that runs k-fold cross-validation and returns mean and standard deviation
def cv_mean_std(model, X, y, k=5):
    s = cross_val_score(model, X, y, cv=k)
    return s.mean(), s.std()

m, sd = cv_mean_std(DecisionTreeClassifier(random_state=1), wine.data, wine.target, k=5)
print("Decision Tree on Wine -> mean:", round(m, 3), "std:", round(sd, 4))

m, sd = cv_mean_std(KNeighborsClassifier(), iris.data, iris.target, k=10)
print("KNN on Iris -> mean:", round(m, 3), "std:", round(sd, 4))
```
