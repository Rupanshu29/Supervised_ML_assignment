### Task 1: ML features in apps I use daily

1. Zomato - shows an estimated delivery time for your order. The model learns from old orders where the real delivery time is known - supervised learning
2. Instagram - the Explore page shows posts similar to what you like. It groups similar posts together without any labels - unsupervised learning
3. Flipkart - decides which banner or deal to show you. If you click, the system gets a reward and learns to show similar things - reinforcement learning

### Task 2: Classify the scenarios

1. Netflix recommending movies based on watch history - supervised. It learns from past data of what the user watched and liked.
2. Spotify grouping similar songs into playlists - unsupervised. Songs have no labels, they are just grouped by similarity.
3. Self-driving car learning to park by trial and error - reinforcement. It gets a reward for good parking and a penalty for mistakes.

### Task 3: ML workflow for Swiggy

```
Data  -->  Preprocessing  -->  Modeling  -->  Evaluation  -->  Deployment
```

1. Data - collect past Swiggy orders with distance, traffic and delivery time.
2. Preprocessing - remove missing values and wrong entries from the data.
3. Modeling - train a model to predict the delivery time.
4. Evaluation - check how close the predicted time is to the real delivery time.
5. Deployment - show the predicted delivery time to users in the Swiggy app.

### Task 4: Spam filter for WhatsApp

In traditional programming, we write fixed rules ourselves, like "if the message has words like free, lottery or winner, mark it as spam". The problem is spammers keep changing their words, so we have to keep updating the rules again and again. In machine learning, we do not write the rules. We give the model many messages already marked as spam or not spam, and it learns the pattern from the data on its own. When new spam comes, we can just train it again with new data, so it adapts much easier than fixed rules.
