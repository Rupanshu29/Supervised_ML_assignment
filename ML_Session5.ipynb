```python
#Task 1: Instagram hours vs battery percentage (scatter plot)
import numpy as np
import matplotlib.pyplot as plt

hours = np.array([1.0, 2.5, 3.0, 4.5, 2.0, 5.5, 3.5])
bat = np.array([72, 55, 48, 30, 62, 18, 42])

plt.scatter(hours, bat, color="blue")
plt.xlabel("Instagram hours per day")
plt.ylabel("Battery % at end of day")
plt.title("Instagram usage vs battery left")
plt.show()
```

```python
#Task 2: Fit simple linear regression and print slope and intercept
from sklearn.linear_model import LinearRegression

x = hours.reshape(-1, 1)
model = LinearRegression()
model.fit(x, bat)

print("Slope (coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
#Task 3: Plot the regression line on the scatter plot
pred = model.predict(x)

plt.scatter(hours, bat, color="blue", label="Actual")
plt.plot(hours, pred, color="red", label="Regression line")
plt.xlabel("Instagram hours per day")
plt.ylabel("Battery % at end of day")
plt.title("Regression line: hours vs battery")
plt.legend()
plt.show()
```

```python
#Task 4: Calculate mean squared error
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(bat, pred)
print("MSE:", mse)

# A lower MSE means the predicted values are closer to the actual values, so the model fits better.
```
