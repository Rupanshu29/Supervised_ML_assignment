```python
#Task 1: Scale IPL player stats with StandardScaler and save to ipl_scaled.csv
import pandas as pd
from sklearn.preprocessing import StandardScaler

ipl = pd.read_csv("ipl_stats.csv")
cols = ["matches", "runs", "wickets", "strike_rate"]

sc = StandardScaler()
scaled = ipl.copy()
scaled[cols] = sc.fit_transform(ipl[cols])
scaled.to_csv("ipl_scaled.csv", index=False)

print(scaled.head())
print("\nMean after scaling:")
print(scaled[cols].mean().round(2) + 0)
print("Std after scaling:")
print(scaled[cols].std(ddof=0).round(2))
```

```python
#Task 2: Scale Flipkart product data between 0 and 1 with MinMaxScaler
from sklearn.preprocessing import MinMaxScaler

fk = pd.read_csv("flipkart_products.csv")

print("Before scaling")
print("Min:")
print(fk.min())
print("Max:")
print(fk.max())

mm = MinMaxScaler()
fk_s = pd.DataFrame(mm.fit_transform(fk), columns=fk.columns)

print("\nAfter scaling")
print("Min:")
print(fk_s.min())
print("Max:")
print(fk_s.max())
```

```python
#Task 3: Pick top 2 features related to popularity using correlation
sp = pd.read_csv("spotify_songs.csv")

corr = sp.corr()["popularity"].drop("popularity")
print(corr)

top2 = corr.abs().sort_values(ascending=False).head(2).index.tolist()
print("\nSelected features:", top2)

# danceability has the highest correlation with popularity (about 0.79), so more danceable songs are usually more popular.
# energy has the second highest correlation (about 0.44), so higher energy songs tend to be somewhat more popular. Tempo is close to 0, so it is left out.
```

```python
#Task 4: Select 3 features with RFE and DecisionTreeRegressor
from sklearn.feature_selection import RFE
from sklearn.tree import DecisionTreeRegressor

zo = pd.read_csv("zomato_restaurants.csv")
X = zo.drop(columns="average_cost_for_two")
y = zo["average_cost_for_two"]

rfe = RFE(DecisionTreeRegressor(random_state=0), n_features_to_select=3)
rfe.fit(X, y)

print("Selected features:", list(X.columns[rfe.support_]))
print("\nRFE ranking (1 means selected):")
for i in range(len(X.columns)):
    print(X.columns[i], "->", rfe.ranking_[i])
```

```python
#Task 5: Detect and remove outliers in user_ratings using IQR
bms = pd.read_csv("bookmyshow_ratings.csv")

q1 = bms["user_ratings"].quantile(0.25)
q3 = bms["user_ratings"].quantile(0.75)
iqr = q3 - q1
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr
print("Q1:", q1, "Q3:", q3, "IQR:", iqr)
print("Allowed range:", low, "to", high)

clean = bms[(bms["user_ratings"] >= low) & (bms["user_ratings"] <= high)]
print("\nRows before:", len(bms))
print("Rows after:", len(clean))
print("Rows dropped:", len(bms) - len(clean))
```
