```python
#Task 1: Gradient Boosting on Iris
# pip install scikit-learn   (run this in a terminal if scikit-learn is not installed)
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier, AdaBoostClassifier
from sklearn.metrics import accuracy_score, classification_report

iris = load_iris()
xtr, xte, ytr, yte = train_test_split(iris.data, iris.target, test_size=0.3, random_state=0)

gb = GradientBoostingClassifier(random_state=0)
gb.fit(xtr, ytr)
print("Test accuracy:", round(accuracy_score(yte, gb.predict(xte)), 3))
```

```python
#Task 2: AdaBoost on Flipkart reviews with classification report
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

rv = pd.read_csv("flipkart_reviews.csv")
xtr2, xte2, ytr2, yte2 = train_test_split(rv["review"], rv["label"], test_size=0.25, random_state=1, stratify=rv["label"])

cv = CountVectorizer(stop_words="english")
xtr_n = cv.fit_transform(xtr2)
xte_n = cv.transform(xte2)

ada = AdaBoostClassifier(n_estimators=200, random_state=1)
ada.fit(xtr_n, ytr2)
print(classification_report(yte2, ada.predict(xte_n)))
```

### Task 3: Weak learner vs strong learner

A weak learner is a very simple model that is only a little better than random guessing. For example, a decision tree with just one question (a decision stump). A rule like "if a post has more than 100 likes, show it on the Explore page" on Instagram is weak, because many good posts have fewer likes and many popular looking posts are boring.

A strong learner is a model that gives high accuracy on its own. For example, the model that Zomato uses to predict the delivery time, which looks at distance, traffic, weather, restaurant preparation time and past orders together, and is usually close to the real time.

Boosting joins many weak learners one after another. Each new learner focuses on the mistakes of the earlier ones, so together they become a strong learner.

```python
#Task 4: Tune n_estimators and learning_rate for Gradient Boosting
combos = [(5, 0.01), (20, 0.05), (50, 0.1), (100, 0.1), (200, 0.5)]

best = 0
best_combo = None
for n, lr in combos:
    m = GradientBoostingClassifier(n_estimators=n, learning_rate=lr, random_state=0)
    m.fit(xtr, ytr)
    acc = accuracy_score(yte, m.predict(xte))
    print("n_estimators =", n, "learning_rate =", lr, "-> accuracy:", round(acc, 3))
    if acc > best:
        best = acc
        best_combo = (n, lr)

print("\nBest combination (n_estimators, learning_rate):", best_combo, "with accuracy", round(best, 3))

# n_estimators 5 with learning_rate 0.01 is far too weak (accuracy 0.6): too few stages with tiny steps cannot learn enough.
# All the other combinations tie at the best accuracy of 0.978, so the first one found, (20, 0.05), is reported.
# Iris is small and easy, so among the ties it is better to choose the simpler and faster model.
```

### Task 5: Boosting use case in an Indian app

Fraud detection in Paytm is a good example. Fraud payments are very rare compared to normal payments, so a simple model misses many of them. Boosting trains many small trees one after another, and each new tree gives more attention to the transactions that the earlier trees got wrong. This helps catch new and unusual fraud patterns, and it also helps keep fewer genuine payments blocked by mistake. Boosting based models like Gradient Boosting and XGBoost are widely used for fraud detection in payment apps for this reason.
