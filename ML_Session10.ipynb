```python
#Task 1: Sigmoid function
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

for v in [-2, 0, 3]:
    print("sigmoid(", v, ") =", sigmoid(v))
```

```python
#Task 2: Logistic regression to predict if an Instagram post goes viral
import pandas as pd
from sklearn.linear_model import LogisticRegression

posts = pd.DataFrame({
    "likes": [120, 450, 300, 5000, 80, 2500, 150, 4200, 700, 3000],
    "has_caption": [0, 1, 0, 1, 0, 1, 1, 0, 1, 1],
    "viral": [0, 0, 0, 1, 0, 1, 0, 1, 0, 1],
})

X = posts[["likes", "has_caption"]]
y = posts["viral"]

lr = LogisticRegression(max_iter=1000)
lr.fit(X, y)

print("Coefficients (likes, has_caption):", lr.coef_[0])
print("Intercept:", lr.intercept_[0])
print("Probability of viral:", lr.predict_proba(X)[:, 1].round(2))
```

```python
#Task 3: Plot sigmoid from -10 to 10 with threshold at 0.5
import matplotlib.pyplot as plt

xs = np.linspace(-10, 10, 200)
plt.plot(xs, sigmoid(xs), color="blue", label="Sigmoid")
plt.axhline(0.5, color="red", linestyle="--", label="Threshold 0.5")
plt.xlabel("x")
plt.ylabel("sigmoid(x)")
plt.title("Sigmoid function")
plt.legend()
plt.show()
```

```python
#Task 4: Classify probability as Viral or Not Viral using a threshold
def classify(prob, threshold=0.5):
    if prob >= threshold:
        return "Viral"
    return "Not Viral"

for p in [0.2, 0.6, 0.8]:
    for t in [0.5, 0.7]:
        print("Probability", p, "threshold", t, "->", classify(p, t))
```
