```python
#Task 1: Load IPL data and count missing values in each column
import pandas as pd

df = pd.read_csv("ipl_players.csv")
print(df.isnull().sum())
```

```python
#Task 2: Fill missing player_age with the median age
med = df["player_age"].median()
df["player_age"] = df["player_age"].fillna(med)

print("Median age:", med)
print(df["player_age"])
```

```python
#Task 3: Replace missing team names with 'Unknown'
df["team"] = df["team"].fillna("Unknown")

print(df.head(10))
```

```python
#Task 4: One-hot encode the venue column
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 20)

enc = pd.get_dummies(df, columns=["venue"], dtype=int)
print(enc.head(10))
print("\nNew columns:", list(enc.columns[-4:]))
```

```python
#Task 5: Label encode the player_role column
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["role_code"] = le.fit_transform(df["player_role"])

print("Mapping from role to code:")
for i in range(len(le.classes_)):
    print(le.classes_[i], "->", i)
print(df[["player_role", "role_code"]].head(10))
```
